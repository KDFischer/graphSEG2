# Interactive plotting and picking of SEG-2 files

This Jupyther notebook shows the simple usage of the python module graphSEG2
(based on https://github.com/tonianr/graphSEG2) to plot SEG-2 files, to pick first onsets and
to analyze the picks with a simple 2 layer (1 itnerface) model.

## Prepare environment

Load all necessary modules, supress user warnings and enable interactive plots in Jupyter notebooks.

```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import warnings
import obspy
from graphSEG2 import SEG2grapher

warnings.filterwarnings(
    "ignore",
    #message=".*WARNING_HEADER.*",
    category=UserWarning
)
```

## Plot SEG-2 file

Read in a SEG-2 file and plot it using a fixed gain for all traces. Manually **upload** your data files and adjust file names below before proceeding.

```python
# define filename and gain
data_file = 'data/12.dat'
pick_file = 'data/picks_12.txt'
gain = 10
```

```python
# plot data
SEG2grapher.plot(data_file, gain)
```

## Pick first onsets

Interactivly pick first onsets using the mouse:

* left click: pick
* right click: remove last pick
* middle click: exit picker

```python
# pick onsets
SEG2grapher.pick(data_file, gain, pick_file)
```

## Analyze data

Use the picked onsets to fit slopes of the direct and refracted waves. Print the velocities of the two layers and its depth.

```python
# analyze picks
m_direct, m_refracted, t_intercept = SEG2grapher.analyze(pick_file, 'false')
```

```python
# calculate and print velocities
v1 = 1/m_direct
v2 = 1/m_refracted
print(f'v_1 = {v1:.2f} km/s, v_2 = {v2:.2f} km/s')

# calculate depth
h = t_intercept / 2.0 * v1*v2/np.sqrt(v2**2-v1**2)
print(f'depth of interface {h:.1f} m')
```
